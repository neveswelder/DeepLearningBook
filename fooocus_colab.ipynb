```python

import os
import subprocess
import sys
import shutil

BASE = "/content"
FOOOCUS = f"{BASE}/Fooocus"
ENV = f"{BASE}/fooocus-env"
PYTHON = f"{ENV}/bin/python"

def run(cmd, cwd=None):
    subprocess.run(cmd, cwd=cwd, check=True)

# 1. Preparar o uv e o Python, se necessário
if not shutil.which("uv"):
    run([sys.executable, "-m", "pip", "install", "-q", "uv"])

if not os.path.isfile(PYTHON):
    run(["uv", "python", "install", "3.10.22"])
    run(["uv", "venv", "--python", "3.10.22", ENV])

# 2. Baixar o Fooocus, se necessário
if not os.path.isdir(os.path.join(FOOOCUS, ".git")):
    if os.path.exists(FOOOCUS):
        shutil.rmtree(FOOOCUS)
    run(["git", "clone", "https://github.com/lllyasviel/Fooocus.git", FOOOCUS])

# 3. Instalar dependências se ainda não estiverem disponíveis
check = subprocess.run(
    [PYTHON, "-c", "import torch, gradio, pygit2"],
    capture_output=True
)
if check.returncode != 0:
    run([PYTHON, "-m", "ensurepip", "--upgrade"])
    run([PYTHON, "-m", "pip", "install", "--upgrade", "pip", "setuptools", "wheel"])
    run([PYTHON, "-m", "pip", "install", "-r", "requirements_versions.txt"], cwd=FOOOCUS)

# 4. Aplicar a versão compatível do MarkupSafe
run([PYTHON, "-m", "pip", "install", "MarkupSafe==2.1.5"])

# 5. Iniciar o Fooocus
os.environ["MPLBACKEND"] = "Agg"
os.chdir(FOOOCUS)
print("Iniciando o Fooocus. Aguarde o link público...")
os.execv(PYTHON, [
    PYTHON,
    "entry_with_update.py",
    "--share",
    "--always-high-vram"
])
```
